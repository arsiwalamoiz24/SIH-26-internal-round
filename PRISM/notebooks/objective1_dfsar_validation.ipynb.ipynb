```python
import numpy as np
import matplotlib.pyplot as plt
from rasterio.enums import Resampling

out_h = 1500
out_w = int(out_h * 24181 / 24794)

def read_overview(path, out_shape):
    with rasterio.open(path) as src:
        arr = src.read(
            1,
            out_shape=(1, out_shape[0], out_shape[1]),
            resampling=Resampling.average
        )
        return arr.squeeze().astype(np.float32)

evn = read_overview(paths["evn"], (out_h, out_w))
vol = read_overview(paths["vol"], (out_h, out_w))
odd = read_overview(paths["odd"], (out_h, out_w))
hlx = read_overview(paths["hlx"], (out_h, out_w))

for name, arr in [("evn", evn), ("vol", vol), ("odd", odd), ("hlx", hlx)]:
    finite = arr[np.isfinite(arr)]
    print(f"{name}: min={finite.min():.4f} max={finite.max():.4f} "
          f"mean={finite.mean():.4f} nan/inf count={(~np.isfinite(arr)).sum()}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

def db_scale(arr):
    # Convert linear power to dB for display: 10*log10(power)
    with np.errstate(divide='ignore', invalid='ignore'):
        db = 10 * np.log10(arr)
    return db

evn_db = db_scale(evn)
vol_db = db_scale(vol)
odd_db = db_scale(odd)
hlx_db = db_scale(hlx)

fig, axes = plt.subplots(2, 2, figsize=(14, 12))
for ax, (name, arr_db) in zip(axes.flat, [("EVN", evn_db), ("VOL", vol_db), ("ODD", odd_db), ("HLX", hlx_db)]):
    finite = arr_db[np.isfinite(arr_db)]
    vmin, vmax = np.percentile(finite, [2, 98])
    im = ax.imshow(arr_db, cmap="viridis", vmin=vmin, vmax=vmax)
    ax.set_title(f"{name} (dB)")
    plt.colorbar(im, ax=ax, shrink=0.7)
plt.tight_layout()
plt.show()
```

```python
total = evn + vol + odd + hlx
total_safe = np.where((total <= 0) | ~np.isfinite(total), np.nan, total)
pv_fraction = vol / total_safe

finite_pv = pv_fraction[np.isfinite(pv_fraction)]
print(f"Pv fraction: min={finite_pv.min():.3f} max={finite_pv.max():.3f} "
      f"mean={finite_pv.mean():.3f} median={np.median(finite_pv):.3f}")
print("Percentiles [10,25,50,75,90]:", np.percentile(finite_pv, [10,25,50,75,90]))

plt.figure(figsize=(10,8))
plt.imshow(pv_fraction, cmap="viridis", vmin=0, vmax=1)
plt.colorbar(label="Volume scattering fraction (Vol / Total)")
plt.title("South Pole — Pv fraction (overview)")
plt.show()
```

```python
p25, p50, p75, p90 = np.percentile(finite_pv, [25, 50, 75, 90])
print(f"p25={p25:.3f} p50={p50:.3f} p75={p75:.3f} p90={p90:.3f}")

score = np.full(pv_fraction.shape, np.nan)
low  = np.isfinite(pv_fraction) & (pv_fraction < p50)
mod  = np.isfinite(pv_fraction) & (pv_fraction >= p50) & (pv_fraction < p90)
high = np.isfinite(pv_fraction) & (pv_fraction >= p90)

score[low]  = 0
score[mod]  = 1
score[high] = 2

plt.figure(figsize=(10,8))
plt.imshow(score, cmap="inferno")
plt.colorbar(label="Tier: 0=below-median, 1=elevated, 2=top-decile Pv")
plt.title("Volume-scattering candidate tiers (data-derived percentiles)")
plt.show()
```

```python
import rasterio
from rasterio.features import geometry_mask
from rasterio.transform import from_bounds
import numpy as np

# Get the transform/bounds we used for our 1500px overview
with rasterio.open(paths["vol"]) as src:
    full_bounds = src.bounds
    full_crs = src.crs

overview_transform = from_bounds(*full_bounds, out_w, out_h)

# Force the PSR CRS to match — since the underlying projection is identical,
# reprojecting will be a no-op (or extremely close); we verify rather than assume
psr_aligned = psr.to_crs(full_crs)

# Rasterize: True where inside a PSR polygon
psr_mask = geometry_mask(
    psr_aligned.geometry,
    out_shape=(out_h, out_w),
    transform=overview_transform,
    invert=True   # True = inside a PSR
)

print("PSR mask pixel count:", psr_mask.sum(), "/", psr_mask.size)
print("Fraction of overview inside a PSR:", psr_mask.mean())

plt.figure(figsize=(10,8))
plt.imshow(psr_mask, cmap="gray")
plt.title("PSR mask rasterized onto DFSAR overview grid")
plt.show()
```

```python
# Combine: PSR mask AND Pv tier
# Reminder: score is 0=below-median, 1=elevated, 2=top-decile Pv (from before)

final_candidate = np.full(pv_fraction.shape, "outside_psr", dtype=object)

inside_psr = psr_mask & np.isfinite(pv_fraction)

final_candidate = np.where(~inside_psr, -1,   # -1 = outside any PSR, not ice-eligible regardless of Pv
                    np.where(score == 2, 2,    # PSR + top-decile Pv = high candidate
                    np.where(score == 1, 1,    # PSR + elevated Pv = moderate candidate
                    0)))                        # PSR + low Pv = low candidate

# Stats: how much of the PSR area falls into each tier
for tier, label in [(-1,"outside PSR"), (0,"PSR, low Pv"), (1,"PSR, moderate Pv"), (2,"PSR, high Pv")]:
    count = (final_candidate == tier).sum()
    print(f"{label}: {count} px")

plt.figure(figsize=(10,8))
cmap = plt.cm.get_cmap("inferno", 4)
im = plt.imshow(final_candidate, cmap=cmap, vmin=-1, vmax=2)
cbar = plt.colorbar(im, ticks=[-1,0,1,2])
cbar.ax.set_yticklabels(["outside PSR", "PSR/low Pv", "PSR/moderate Pv", "PSR/high Pv"])
plt.title("PSR-gated volume-scattering candidate map (South Pole overview)")
plt.show()
```

```python
from rasterio.windows import from_bounds
from rasterio.features import geometry_mask as geom_mask

shortlist_ids = [
    "SP_832640_0090770", "SP_830080_0535120", "SP_842420_0421060",
    "SP_817950_1586580", "SP_840980_0797630", "SP_819860_1568660",
    "SP_809570_2454450"
]

def read_full_res_window(path, bounds):
    with rasterio.open(path) as src:
        window = from_bounds(*bounds, transform=src.transform)
        arr = src.read(1, window=window)
        win_transform = src.window_transform(window)
    return arr.astype(np.float32), win_transform

crater_results = {}

for psr_id in shortlist_ids:
    row = psr_aligned[psr_aligned.PSR_ID == psr_id].iloc[0]
    geom = row.geometry
    minx, miny, maxx, maxy = geom.bounds
    buffer = 1000  # meters of context around the PSR
    bounds = (minx - buffer, miny - buffer, maxx + buffer, maxy + buffer)

    evn_fr, tr = read_full_res_window(paths["evn"], bounds)
    vol_fr, _  = read_full_res_window(paths["vol"], bounds)
    odd_fr, _  = read_full_res_window(paths["odd"], bounds)
    hlx_fr, _  = read_full_res_window(paths["hlx"], bounds)

    total_fr = evn_fr + vol_fr + odd_fr + hlx_fr
    valid_fr = np.isfinite(total_fr) & (total_fr > 0)
    total_safe_fr = np.where(valid_fr, total_fr, np.nan)
    pv_fr = vol_fr / total_safe_fr

    psr_poly_mask = geom_mask([geom], out_shape=evn_fr.shape, transform=tr, invert=True)

    crater_results[psr_id] = dict(
        pv=pv_fr, valid=valid_fr, psr_mask=psr_poly_mask,
        shape=evn_fr.shape, area_km2=row.area_km2 if hasattr(row, 'area_km2') else row.area
    )

    inside = psr_poly_mask & valid_fr
    n = inside.sum()
    if n > 0:
        print(f"{psr_id}: window {evn_fr.shape}, {n} valid px inside PSR, "
              f"full-res mean Pv={pv_fr[inside].mean():.3f}, "
              f"median Pv={np.median(pv_fr[inside]):.3f}")
    else:
        print(f"{psr_id}: window {evn_fr.shape}, NO valid radar data inside PSR at full res")
```

```python
for psr_id, d in crater_results.items():
    pv, valid, psr_mask = d["pv"], d["valid"], d["psr_mask"]

    inside_psr = psr_mask & valid
    outside_psr = (~psr_mask) & valid   # same local window, but NOT in the PSR polygon

    n_in, n_out = inside_psr.sum(), outside_psr.sum()
    if n_in == 0 or n_out == 0:
        print(f"{psr_id}: insufficient data for comparison (in={n_in}, out={n_out})")
        continue

    mean_in = pv[inside_psr].mean()
    mean_out = pv[outside_psr].mean()
    print(f"{psr_id}: PSR mean Pv={mean_in:.3f} (n={n_in})  |  "
          f"local surroundings mean Pv={mean_out:.3f} (n={n_out})  |  "
          f"difference={mean_in - mean_out:+.3f}")
```
